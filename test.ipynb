```python
import h5py
import numpy as np

file_path = "/vePFS-0x0d/DailyEEG/de_output/sub_03_de.h5"

EXPECTED_SAMPLE_PER_TRIAL = 23
EXPECTED_EEG_LEN = 2000  # 10s × 200Hz

def check_h5(file_path):
    print(f"\n========== 检查文件: {file_path} ==========")

    error_count = 0

    with h5py.File(file_path, "r") as f:
        trials = list(f.keys())

        print(f"总 trial 数: {len(trials)}")

        for trial_name in trials:
            trial = f[trial_name]
            samples = list(trial.keys())

            print(f"\n--- {trial_name} ---")

            # ===== 1. sample数量检查 =====
            if len(samples) != EXPECTED_SAMPLE_PER_TRIAL:
                print(f"❌ sample数量错误: {len(samples)} (应为 {EXPECTED_SAMPLE_PER_TRIAL})")
                error_count += 1
            else:
                print(f"✅ sample数量正确: {len(samples)}")

            # ===== 2. label一致性 =====
            labels = []

            for sample_name in samples:
                sample = trial[sample_name]

                # ===== attrs检查 =====
                for attr in ["label", "sub_id", "day_id"]:
                    if attr not in sample.attrs:
                        print(f"❌ {trial_name}/{sample_name} 缺少属性: {attr}")
                        error_count += 1

                labels.append(sample.attrs.get("label", None))

                # ===== DE检查 =====
                if "de" not in sample:
                    print(f"❌ 缺少de: {trial_name}/{sample_name}")
                    error_count += 1
                else:
                    de = sample["de"][:]
                    if np.isnan(de).any() or np.isinf(de).any():
                        print(f"❌ DE存在NaN/Inf: {trial_name}/{sample_name}")
                        error_count += 1

                # ===== EEG检查 =====
                if "raw_eeg_10s_window" not in sample:
                    print(f"❌ 缺少EEG: {trial_name}/{sample_name}")
                    error_count += 1
                else:
                    eeg = sample["raw_eeg_10s_window"]

                    if eeg.shape[1] != EXPECTED_EEG_LEN:
                        print(f"❌ EEG长度错误: {eeg.shape}")
                        error_count += 1

                    data = eeg[:]
                    if np.isnan(data).any() or np.isinf(data).any():
                        print(f"❌ EEG存在NaN/Inf: {trial_name}/{sample_name}")
                        error_count += 1

            # ===== label一致性检查 =====
            if len(labels) > 0:
                first_label = labels[0]
                for i, lb in enumerate(labels):
                    if not np.array_equal(lb, first_label):
                        print(f"❌ label不一致: sample{i}")
                        error_count += 1
                        break
                else:
                    print("✅ label一致")

    print("\n========== 检查完成 ==========")
    print(f"总错误数: {error_count}")

    if error_count == 0:
        print("🎉 数据完全正确！可以放心训练模型")
    else:
        print("⚠️ 存在问题，请检查数据")


if __name__ == "__main__":
    check_h5(file_path)
```

```python
import os
import h5py
import glob
#DATA_OUT = r"/vePFS-0x0d/DailyEEG/de_output"
DATA_OUT = r"/vePFS-0x0d/DailyEEG/de_output_addtime_new/"
h5_files = sorted(glob.glob(os.path.join(DATA_OUT, "sub_*.h5")))

all_total_samples = 0  # 全局所有被试总样本数

for h5_path in h5_files:
    fname = os.path.basename(h5_path)
    print(f"\n===== File: {fname} =====")
    sub_total_trial = 0
    sub_total_sample = 0
    trial_info = {}

    with h5py.File(h5_path, "r") as f:
        trial_groups = [g for g in f.keys() if g.startswith("trial")]
        sub_total_trial = len(trial_groups)
        print(f"Total trials in file: {sub_total_trial}")

        for trial_name in trial_groups:
            trial_num = int(trial_name.replace("trial", ""))
            sample_groups = [s for s in f[trial_name].keys() if s.startswith("sample")]
            sample_cnt = len(sample_groups)
            trial_info[trial_num] = sample_cnt
            sub_total_sample += sample_cnt
            print(f"  {trial_name}: {sample_cnt} samples")

    print(f"Subject total samples: {sub_total_sample}")
    all_total_samples += sub_total_sample

# 全部文件统计汇总
print("\n========================================")
print(f"All subjects total samples sum: {all_total_samples}")
print("========================================")
```

```python
#!/usr/bin/env python3
"""
统计每个联合域 ID 对应的真实 sub_id 和 day_id。
新增1：统计每个 day_id 下各 sub_id 样本数量
新增2：统计每个 sub_id 包含哪些 day_id、每日样本及总样本
"""

import json
import os
import numpy as np


def main():
    #data_dir = "/vePFS-0x0d/DailyEEG/processed_cadt"
    
    data_dir = "/vePFS-0x0d/home/zjm/processed_cadt_addtime_new"

    # 加载数据
    d_path = os.path.join(data_dir, "d.npy")
    sub_path = os.path.join(data_dir, "sub.npy")
    
    if not os.path.exists(d_path) or not os.path.exists(sub_path):
        print(f"错误：未找到 {d_path} 或 {sub_path}")
        return
    
    all_d = np.load(d_path)
    all_sub = np.load(sub_path)
    
    # 从 meta.json 读取配置
    meta_path = os.path.join(data_dir, "meta.json")
    domains_per_subject = 10
    if os.path.exists(meta_path):
        with open(meta_path, "r", encoding="utf-8") as f:
            meta = json.load(f)
            domains_per_subject = meta.get("domains_per_subject", 10)
    
    print(f"域 ID 计算公式: domain_id = (sub_id - 1) * {domains_per_subject} + day_id")
    print(f"数据集中共有 {len(all_d)} 个样本")
    print(f"唯一域 ID 数量: {len(np.unique(all_d))}")
    print(f"唯一被试数量: {len(np.unique(all_sub))}")
    print()
    
    # 1. 原有：每个域ID映射信息
    domain_mapping = {}
    # 2. day -> {sub_id: count}
    day_sub_stats = {}
    # 3. sub -> {day_id: count} 本次新增
    sub_day_stats = {}
    
    for d, sub in zip(all_d, all_sub):
        sub = int(sub)
        if d not in domain_mapping:
            # 反向计算 sub_id 和 day_id
            sub_id_calc = d // domains_per_subject + 1
            day_id_calc = d % domains_per_subject
            domain_mapping[d] = {
                "sub_id": sub_id_calc,
                "day_id": day_id_calc,
                "actual_sub_id": sub,
                "sample_count": 0
            }
        domain_mapping[d]["sample_count"] += 1
        
        day_id = domain_mapping[d]["day_id"]
        
        # 统计 day -> sub
        if day_id not in day_sub_stats:
            day_sub_stats[day_id] = {}
        if sub not in day_sub_stats[day_id]:
            day_sub_stats[day_id][sub] = 0
        day_sub_stats[day_id][sub] += 1
        
        # 统计 sub -> day（新增）
        if sub not in sub_day_stats:
            sub_day_stats[sub] = {}
        if day_id not in sub_day_stats[sub]:
            sub_day_stats[sub][day_id] = 0
        sub_day_stats[sub][day_id] += 1
    
    # 输出原有域ID明细表格
    print("=" * 70)
    print(f"{'域ID':<8} {'sub_id(计算)':<15} {'day_id':<10} {'sub_id(真实)':<15} {'样本数':<8}")
    print("-" * 70)
    for domain_id in sorted(domain_mapping.keys()):
        info = domain_mapping[domain_id]
        print(f"{domain_id:<8} {info['sub_id']:<15} {info['day_id']:<10} {info['actual_sub_id']:<15} {info['sample_count']:<8}")
    print("=" * 70)
    print()
    
    # 验证计算的 sub_id 是否与真实 sub_id 一致
    mismatches = 0
    for d, info in domain_mapping.items():
        if info["sub_id"] != info["actual_sub_id"]:
            mismatches += 1
            print(f"警告：域ID {d} 的计算 sub_id({info['sub_id']}) 与真实 sub_id({info['actual_sub_id']}) 不一致")
    if mismatches == 0:
        print("✓ 所有域ID的映射关系验证通过")
    else:
        print(f"✗ 发现 {mismatches} 个映射不一致")
    print()
    
    # 第一部分：按 day_id 分组，各 sub 样本
    print("=" * 60)
    print("【统计1：按 day_id 分组，各 sub_id 样本数量】")
    print("=" * 60)
    for day_id in sorted(day_sub_stats.keys()):
        sub_dict = day_sub_stats[day_id]
        print(f"\nDay ID = {day_id}")
        print(f"{'sub_id':<10} {'样本数量':<10}")
        print("-" * 20)
        total_day = 0
        for sid in sorted(sub_dict.keys()):
            cnt = sub_dict[sid]
            total_day += cnt
            print(f"{sid:<10} {cnt:<10}")
        print(f"当日总样本数: {total_day}")
    print()
    
    # 第二部分：按 sub_id 分组，拥有多少个 day_id + 每日样本（本次新增）
    print("=" * 60)
    print("【统计2：按 sub_id 分组，拥有的 day_id 及对应样本】")
    print("=" * 60)
    for sub_id in sorted(sub_day_stats.keys()):
        day_dict = sub_day_stats[sub_id]
        unique_day_num = len(day_dict)
        total_sub_samples = sum(day_dict.values())
        print(f"\n被试 sub_id = {sub_id} | 拥有不同 day 数量：{unique_day_num} | 该被试总样本：{total_sub_samples}")
        print(f"{'day_id':<10} {'当日样本数':<10}")
        print("-" * 22)
        for did in sorted(day_dict.keys()):
            cnt = day_dict[did]
            print(f"{did:<10} {cnt:<10}")


if __name__ == "__main__":
    main()
```

```python
"""
DailyEEG 数据集统计分析脚本

功能：
1. 标签统计：11种情绪的均值、标准差、分布区间
2. 标签相关性：情绪之间的皮尔逊相关系数矩阵
3. 特征统计：训练集特征的均值、标准差、最大最小值、范数分布
"""

import numpy as np
import os
import matplotlib
matplotlib.use('Agg')  # 使用非交互式后端
import matplotlib.pyplot as plt
from matplotlib import cm

# 情绪标签名称（假设顺序）
EMOTION_LABELS = [
    "inspired", "alert", "determined", "attentive", "active",
    "hostile", "nervous", "upset", "afraid", "ashamed", "fatigue",
]


def load_data(data_dir="/vePFS-0x0e/DailyEEG/processed_cadt"):
    """加载DailyEEG数据集"""
    x_path = os.path.join(data_dir, "X.npy")
    y_path = os.path.join(data_dir, "y.npy")
    d_path = os.path.join(data_dir, "d.npy")
    
    print(f"加载数据: {data_dir}")
    print(f"  X.npy: {os.path.exists(x_path)}")
    print(f"  y.npy: {os.path.exists(y_path)}")
    print(f"  d.npy: {os.path.exists(d_path)}")
    
    all_x = np.load(x_path).astype(np.float32)
    all_y = np.load(y_path).astype(np.float32)
    all_d = np.load(d_path).astype(np.int64)
    
    if all_y.ndim == 1:
        all_y = all_y.reshape(-1, 1)
    
    print(f"\n数据形状:")
    print(f"  X: {all_x.shape} (样本数, 时间点, 通道数)")
    print(f"  y: {all_y.shape} (样本数, 情绪维度)")
    print(f"  d: {all_d.shape} (样本数,)")
    print(f"  域范围: [{all_d.min()}, {all_d.max()}]")
    print(f"  域数量: {len(np.unique(all_d))}")
    
    return all_x, all_y, all_d


def compute_label_statistics(y):
    """计算标签统计信息"""
    n_emotions = y.shape[1]
    
    print("\n" + "="*60)
    print("标签统计 (11种情绪评分, 范围1-5)")
    print("="*60)
    
    # 均值、标准差、最小值、最大值、中位数
    means = np.mean(y, axis=0)
    stds = np.std(y, axis=0)
    mins = np.min(y, axis=0)
    maxs = np.max(y, axis=0)
    medians = np.median(y, axis=0)
    ranges = maxs - mins
    
    print(f"\n{'情绪':<15} {'均值':>8} {'标准差':>8} {'最小值':>8} {'最大值':>8} {'中位数':>8} {'范围':>8}")
    print("-" * 75)
    for i in range(n_emotions):
        name = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"Emotion_{i}"
        print(f"{name:<15} {means[i]:>8.4f} {stds[i]:>8.4f} {mins[i]:>8.4f} {maxs[i]:>8.4f} {medians[i]:>8.4f} {ranges[i]:>8.4f}")
    
    # 每个情绪评分的分布（1-5分的计数）
    print("\n" + "-"*60)
    print("情绪评分分布 (1-5分的样本数)")
    print("-"*60)
    
    score_bins = np.arange(0.5, 5.6, 1.0)  # [0.5, 1.5, 2.5, 3.5, 4.5, 5.5]
    
    print(f"\n{'情绪':<15}", end="")
    for s in range(1, 6):
        print(f"  {s}分", end="")
    print("    样本数")
    
    for i in range(n_emotions):
        name = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"Emotion_{i}"
        counts, _ = np.histogram(y[:, i], bins=score_bins)
        print(f"{name:<15}", end="")
        for c in counts:
            print(f"{c:>6}", end="")
        print(f"{len(y):>10}")
    
    return means, stds, mins, maxs, medians


def compute_label_correlation(y):
    """计算标签之间的相关性矩阵"""
    n_emotions = y.shape[1]
    
    print("\n" + "="*60)
    print("情绪相关性矩阵 (皮尔逊相关系数)")
    print("="*60)
    
    # 计算皮尔逊相关系数
    corr_matrix = np.corrcoef(y.T)
    
    # 打印相关性矩阵
    print("\n", end="")
    print(f"{'':<15}", end="")
    for i in range(n_emotions):
        name = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"E{i}"
        print(f"{name[:4]:>8}", end="")
    print()
    
    for i in range(n_emotions):
        name_i = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"E{i}"
        print(f"{name_i[:15]:<15}", end="")
        for j in range(n_emotions):
            print(f"{corr_matrix[i, j]:>8.4f}", end="")
        print()
    
    # 找出高度相关的情绪对
    print("\n高度相关的情绪对 (|r| > 0.7):")
    print("-" * 50)
    for i in range(n_emotions):
        for j in range(i + 1, n_emotions):
            corr = corr_matrix[i, j]
            if abs(corr) > 0.7:
                name_i = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"E{i}"
                name_j = EMOTION_LABELS[j] if j < len(EMOTION_LABELS) else f"E{j}"
                print(f"  {name_i:<12} - {name_j:<12}: r = {corr:.4f} {'正相关' if corr > 0 else '负相关'}")
    
    return corr_matrix


def compute_feature_statistics(x):
    """计算特征统计信息（使用归一化后的数据）"""
    print("\n" + "="*60)
    print("特征统计 (归一化后)")
    print("="*60)
    
    # 对每个样本进行归一化（和dataset中的处理一致）
    def norm(z):
        return (z - np.mean(z, axis=0)) / (np.std(z, axis=0) + 1e-6)
    
    n_samples, n_timepoints, n_channels = x.shape
    
    # 计算每个通道的统计
    print(f"\n数据形状: {x.shape}")
    print(f"样本数: {n_samples}, 时间点: {n_timepoints}, 通道数: {n_channels}")
    
    # 归一化后的特征
    x_norm = np.zeros_like(x)
    for i in range(n_samples):
        x_norm[i] = norm(x[i])
    
    # 全局统计
    print("\n全局统计:")
    print(f"  均值: {np.mean(x_norm):.6f}")
    print(f"  标准差: {np.std(x_norm):.6f}")
    print(f"  最小值: {np.min(x_norm):.6f}")
    print(f"  最大值: {np.max(x_norm):.6f}")
    
    # 按通道统计
    print("\n各通道统计:")
    print(f"{'通道':>6} {'均值':>10} {'标准差':>10} {'最小值':>10} {'最大值':>10}")
    print("-" * 50)
    for ch in range(n_channels):
        ch_data = x_norm[:, :, ch]
        print(f"{ch:>6} {np.mean(ch_data):>10.6f} {np.std(ch_data):>10.6f} "
              f"{np.min(ch_data):>10.6f} {np.max(ch_data):>10.6f}")
    
    # 计算特征范数分布（每个样本的L2范数）
    print("\n特征范数分布:")
    sample_norms = np.linalg.norm(x_norm.reshape(n_samples, -1), axis=1)
    print(f"  范数均值: {np.mean(sample_norms):.4f}")
    print(f"  范数标准差: {np.std(sample_norms):.4f}")
    print(f"  范数最小值: {np.min(sample_norms):.4f}")
    print(f"  范数最大值: {np.max(sample_norms):.4f}")
    print(f"  范数中位数: {np.median(sample_norms):.4f}")
    
    return x_norm, sample_norms


def plot_distributions(y, sample_norms, corr_matrix, output_dir="stats_plots"):
    """绘制统计分布图"""
    os.makedirs(output_dir, exist_ok=True)
    n_emotions = y.shape[1]
    
    # 1. 情绪评分直方图
    plt.figure(figsize=(15, 10))
    for i in range(n_emotions):
        plt.subplot(3, 4, i + 1)
        name = EMOTION_LABELS[i] if i < len(EMOTION_LABELS) else f"Emotion_{i}"
        plt.hist(y[:, i], bins=np.arange(0.5, 5.6, 1.0), edgecolor='black', alpha=0.7)
        plt.title(name)
        plt.xlabel('Score')
        plt.ylabel('Count')
        plt.xlim(0.5, 5.5)
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'emotion_histograms.png'), dpi=150)
    plt.close()
    print(f"  已保存: {os.path.join(output_dir, 'emotion_histograms.png')}")
    
    # 2. 情绪相关性热力图
    plt.figure(figsize=(10, 8))
    plt.imshow(corr_matrix, cmap='RdBu_r', vmin=-1, vmax=1)
    plt.colorbar()
    emotion_names_short = [name[:6] for name in EMOTION_LABELS[:n_emotions]]
    plt.xticks(range(n_emotions), emotion_names_short, rotation=45, ha='right')
    plt.yticks(range(n_emotions), emotion_names_short)
    plt.title('Emotion Correlation Matrix')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'emotion_correlation.png'), dpi=150)
    plt.close()
    print(f"  已保存: {os.path.join(output_dir, 'emotion_correlation.png')}")
    
    # 3. 特征范数分布
    plt.figure(figsize=(10, 6))
    plt.hist(sample_norms, bins=50, edgecolor='black', alpha=0.7)
    plt.axvline(np.mean(sample_norms), color='red', linestyle='--', label=f'均值: {np.mean(sample_norms):.2f}')
    plt.axvline(np.median(sample_norms), color='green', linestyle='--', label=f'中位数: {np.median(sample_norms):.2f}')
    plt.title('Feature Norm Distribution')
    plt.xlabel('L2 Norm')
    plt.ylabel('Count')
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'feature_norm_distribution.png'), dpi=150)
    plt.close()
    print(f"  已保存: {os.path.join(output_dir, 'feature_norm_distribution.png')}")
    
    # 4. 情绪评分箱线图
    plt.figure(figsize=(12, 6))
    plt.boxplot(y)
    plt.xticks(range(1, n_emotions + 1), EMOTION_LABELS[:n_emotions], rotation=45, ha='right')
    plt.ylabel('Score')
    plt.title('Emotion Score Distribution')
    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, 'emotion_boxplot.png'), dpi=150)
    plt.close()
    print(f"  已保存: {os.path.join(output_dir, 'emotion_boxplot.png')}")


def main():
    # 加载数据
    x, y, d = load_data()
    
    # 计算标签统计
    means, stds, mins, maxs, medians = compute_label_statistics(y)
    
    # 计算标签相关性
    corr_matrix = compute_label_correlation(y)
    
    # 计算特征统计（只使用前1000个样本加速）
    sample_size = min(1000, len(x))
    print(f"\n使用 {sample_size} 个样本计算特征统计...")
    x_norm, sample_norms = compute_feature_statistics(x[:sample_size])
    
    # 绘制分布图
    print("\n绘制统计分布图...")
    plot_distributions(y, sample_norms, corr_matrix)
    
    print("\n" + "="*60)
    print("统计分析完成!")
    print("="*60)


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
验证域ID与被试ID的映射关系（修复版）

使用动态累积偏移方案正确反推 sub_id 和 day_id：
- 从 meta.json 读取 domain_offset_map 和 domains_per_subject_map
- 根据 domain_id 查找对应的 sub_id 和 day_id
- domain_id = offset + day_id
"""

import argparse
import json
import os

import numpy as np


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(description="验证域ID与被试ID的映射关系")
    parser.add_argument(
        "--data-dir",
        type=str,
        default=" /vePFS-0x0d/home/zjm/processed_cadt_addtime_new",
        help="CADT 对齐后的 NPY 数据目录",
    )
    return parser


def main() -> None:
    parser = parse_args()
    # 使用 parse_known_args() 忽略 ipykernel 注入的 -f 参数
    args, _ = parser.parse_known_args()
    
    # 加载数据
    d_path = os.path.join(args.data_dir, "d.npy")
    sub_path = os.path.join(args.data_dir, "sub.npy")
    meta_path = os.path.join(args.data_dir, "meta.json")
    
    if not os.path.exists(d_path) or not os.path.exists(sub_path):
        raise FileNotFoundError(f"未在 {args.data_dir} 找到 d.npy 或 sub.npy")
    
    all_d = np.load(d_path)
    all_sub = np.load(sub_path)
    
    # 加载元数据
    domain_offset_map: dict[int, int] = {}
    domains_per_subject_map: dict[int, int] = {}
    
    if os.path.exists(meta_path):
        with open(meta_path, "r", encoding="utf-8") as f:
            meta = json.load(f)
            # 读取动态域配置
            if "domain_offset_map" in meta:
                domain_offset_map = {int(k): v for k, v in meta["domain_offset_map"].items()}
            if "domains_per_subject_map" in meta:
                domains_per_subject_map = {int(k): v for k, v in meta["domains_per_subject_map"].items()}
    
    print(f"数据集中共有 {len(all_d)} 个样本")
    print(f"唯一域 ID 数量: {len(np.unique(all_d))}")
    print(f"唯一被试数量: {len(np.unique(all_sub))}")
    print()
    
    # 打印动态域配置
    print("动态域配置:")
    for sub_id in sorted(domain_offset_map.keys()):
        start = domain_offset_map[sub_id]
        end = start + domains_per_subject_map.get(sub_id, 10) - 1
        print(f"  sub-{sub_id:02d}: offset={start}, domains_per_subject={domains_per_subject_map.get(sub_id, 10)}, 域范围 [{start}, {end}]")
    print()
    
    # 1. 每个域ID映射信息
    domain_mapping = {}
    # 2. day -> {sub_id: count}
    day_sub_stats = {}
    # 3. sub -> {day_id: count}
    sub_day_stats = {}
    
    # 预排序偏移量用于快速查找
    sorted_sub_ids = sorted(domain_offset_map.keys())
    sorted_offsets = [domain_offset_map[sub_id] for sub_id in sorted_sub_ids]
    
    def find_sub_and_day(domain_id: int) -> tuple[int, int]:
        """根据 domain_id 查找对应的 sub_id 和 day_id"""
        # 找到最大的 offset <= domain_id
        for i in range(len(sorted_offsets)-1, -1, -1):
            if sorted_offsets[i] <= domain_id:
                sub_id = sorted_sub_ids[i]
                day_id = domain_id - sorted_offsets[i]
                return sub_id, day_id
        return 0, domain_id  # 默认回退
    
    for d, sub in zip(all_d, all_sub):
        sub = int(sub)
        if d not in domain_mapping:
            # 使用动态偏移正确反推 sub_id 和 day_id
            sub_id_calc, day_id_calc = find_sub_and_day(int(d))
            domain_mapping[d] = {
                "sub_id": sub_id_calc,
                "day_id": day_id_calc,
                "actual_sub_id": sub,
                "sample_count": 0
            }
        domain_mapping[d]["sample_count"] += 1
        
        day_id = domain_mapping[d]["day_id"]
        
        # 统计 day -> sub
        if day_id not in day_sub_stats:
            day_sub_stats[day_id] = {}
        if sub not in day_sub_stats[day_id]:
            day_sub_stats[day_id][sub] = 0
        day_sub_stats[day_id][sub] += 1
        
        # 统计 sub -> day
        if sub not in sub_day_stats:
            sub_day_stats[sub] = {}
        if day_id not in sub_day_stats[sub]:
            sub_day_stats[sub][day_id] = 0
        sub_day_stats[sub][day_id] += 1
    
    # 输出域ID明细表格
    print("=" * 70)
    print(f"{'域ID':<8} {'sub_id(计算)':<15} {'day_id':<10} {'sub_id(真实)':<15} {'样本数':<8}")
    print("-" * 70)
    for domain_id in sorted(domain_mapping.keys()):
        info = domain_mapping[domain_id]
        print(f"{domain_id:<8} {info['sub_id']:<15} {info['day_id']:<10} {info['actual_sub_id']:<15} {info['sample_count']:<8}")
    print("=" * 70)
    
    # 检查映射一致性
    mismatches = []
    for domain_id, info in domain_mapping.items():
        if info["sub_id"] != info["actual_sub_id"]:
            mismatches.append({
                "domain_id": domain_id,
                "calculated_sub": info["sub_id"],
                "actual_sub": info["actual_sub_id"],
                "day_id": info["day_id"],
                "sample_count": info["sample_count"]
            })
    
    if mismatches:
        print(f"\n警告：发现 {len(mismatches)} 个映射不一致")
        for m in mismatches:
            print(f"  域ID {m['domain_id']}: 计算 sub_id({m['calculated_sub']}) 与真实 sub_id({m['actual_sub']}) 不一致，样本数={m['sample_count']}")
    else:
        print("\n✓ 所有域ID映射一致")


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
排查丢失的域ID问题：检查day_id分配是否正确
"""

import h5py
import numpy as np
import glob
import os

INPUT_DIR = "/vePFS-0x0d/DailyEEG/de_output_addtime_new"

def build_domain_id(sub_id: int, day_id: int, domains_per_subject: int = 10) -> int:
    return (sub_id - 1) * domains_per_subject + day_id

missing_domains = [19, 49, 98, 99, 119, 138, 139, 149]

print("=" * 80)
print("丢失域ID分析")
print("=" * 80)

for domain in missing_domains:
    sub_id = (domain // 10) + 1
    expected_day_id = domain % 10
    print(f"\n域ID {domain} → sub-{sub_id:02d}, 期望day_id={expected_day_id}")

print("\n" + "=" * 80)
print("实际H5文件中的day_id分布")
print("=" * 80)

h5_files = sorted(glob.glob(os.path.join(INPUT_DIR, "sub_*_de.h5")))

for h5_path in h5_files:
    sub_id = int(os.path.basename(h5_path).split("_")[1])
    day_ids = []
    
    with h5py.File(h5_path, "r") as hf:
        trial_keys = sorted(hf.keys(), key=lambda k: int(k.replace("trial", "")))
        for trial_key in trial_keys:
            trial_grp = hf[trial_key]
            sample_keys = sorted(trial_grp.keys(), key=lambda k: int(k.replace("sample", "")))
            if sample_keys:
                sample_grp = trial_grp[sample_keys[0]]
                day_ids.append(int(sample_grp.attrs["day_id"]))
    
    unique_days = sorted(set(day_ids))
    print(f"sub-{sub_id:02d}: day_ids={unique_days}, 最大day_id={max(day_ids) if day_ids else 'N/A'}")
    
    # 检查是否缺少期望的day_id
    expected_days = set(range(10))  # 0-9
    actual_days = set(unique_days)
    missing = sorted(expected_days - actual_days)
    if missing:
        print(f"  ⚠️ 缺少day_id: {missing}")
        for d in missing:
            domain_id = build_domain_id(sub_id, d)
            if domain_id in missing_domains:
                print(f"    → 对应丢失域ID: {domain_id}")

print("\n" + "=" * 80)
print("总结：问题根源")
print("=" * 80)
print("""
问题在于 test_load.py 中 day_id 的计算方式：

1. day_id 通过 np.searchsorted 基于 sample_num_each_day 列表计算
2. 当某个session没有有效样本时，不会往 sample_num_each_day 追加记录
3. 后续session的day_id因此被错误地"前移"了

例如 sub-02：
- 实际session: ses-01(day0), ses-03(day2), ses-04(day3), ..., ses-10(day9)
- 但 sample_num_each_day 只有9个元素
- 导致 ses-10 被分配为 day8，而不是 day9
- 最终域ID 19=(2-1)*10+9 丢失

修复方案：
需要修改 day_id 的计算逻辑，基于实际的session编号来分配day_id，
而不是基于有效样本数量的累计索引。
""")
```

```python
#!/usr/bin/env python3
"""
查看 de_output_addtime_new 目录中H5文件的实际域ID分布
"""

import glob
import os
import h5py
import numpy as np


def main():
    #h5_dir = "/vePFS-0x0d/DailyEEG/de_output_addtime_new"
    h5_dir = "/vePFS-0x0d/home/zjm/de_output_addtime_new"
    h5_files = sorted(glob.glob(os.path.join(h5_dir, "sub_*_de.h5")))
    
    if not h5_files:
        print(f"未在 {h5_dir} 找到 H5 文件")
        return
    
    print("=" * 80)
    print(f"查看 {h5_dir} 中的域ID分布")
    print("=" * 80)
    
    all_domains = {}  # (sub_id, day_id) -> sample_count
    
    for h5_path in h5_files:
        sub_name = os.path.basename(h5_path)
        sub_id = int(sub_name.replace("sub_", "").replace("_de.h5", ""))
        
        with h5py.File(h5_path, "r") as hf:
            trial_keys = sorted(hf.keys(), key=lambda k: int(k.replace("trial", "")))
            
            for trial_key in trial_keys:
                trial_grp = hf[trial_key]
                sample_keys = sorted(trial_grp.keys(), key=lambda k: int(k.replace("sample", "")))
                
                for sample_key in sample_keys:
                    sample_grp = trial_grp[sample_key]
                    if "raw_eeg_10s_window" not in sample_grp:
                        continue
                    day_id = int(sample_grp.attrs["day_id"])
                    key = (sub_id, day_id)
                    if key not in all_domains:
                        all_domains[key] = 0
                    all_domains[key] += 1
    
    # 按 sub_id 和 day_id 排序输出
    print(f"\n{'sub_id':<8} {'day_id':<8} {'session':<12} {'样本数':<8}")
    print("-" * 40)
    
    for (sub_id, day_id), count in sorted(all_domains.items()):
        print(f"{sub_id:<8} {day_id:<8} ses-{day_id+1:02d}    {count:<8}")
    
    print("\n" + "=" * 80)
    print("空的域（有session目录但无样本）：")
    print("-" * 40)
    
    # 检查每个被试的实际session目录
    eeg_deriv_root = "/vePFS-0x0d/DailyEEG/derivatives/eeg-prep"
    sub_dirs = sorted(glob.glob(os.path.join(eeg_deriv_root, "sub-*")))
    
    empty_domains = []
    for sub_dir in sub_dirs:
        sub_id = int(os.path.basename(sub_dir).replace("sub-", ""))
        ses_dirs = sorted(glob.glob(os.path.join(sub_dir, "ses-*")))
        
        for ses_dir in ses_dirs:
            ses_id = os.path.basename(ses_dir)
            day_id = int(ses_id.replace("ses-", "")) - 1
            
            if (sub_id, day_id) not in all_domains:
                empty_domains.append((sub_id, day_id, ses_id))
    
    for sub_id, day_id, ses_id in sorted(empty_domains):
        print(f"sub-{sub_id:02d} {ses_id} (day_id={day_id})：无样本")
    
    print(f"\n共有 {len(all_domains)} 个有样本的域，{len(empty_domains)} 个空域")


if __name__ == "__main__":
    main()
```

```python
"""
查看 processed_cadt_addtime_new 中样本的索引信息
"""
import numpy as np

output_dir = "/vePFS-0x0d/DailyEEG/processed_cadt_addtime_new"

# 加载数据
d = np.load(f"{output_dir}/d.npy")  # 域 ID
ts = np.load(f"{output_dir}/ts.npy")  # 时间戳
sub = np.load(f"{output_dir}/sub.npy")  # 被试编号

print(f"总样本数: {len(d)}")
print(f"域 ID 范围: [{d.min()}, {d.max()}]")
print(f"时间戳范围: [{ts.min():.1f}, {ts.max():.1f}] 秒")

# 查看每个域的样本数量
unique_domains = np.unique(d)
print(f"\n域数量: {len(unique_domains)}")

# 选几个代表性的域查看详细信息
sample_domains = unique_domains[:5]
for domain_id in sample_domains:
    mask = d == domain_id
    indices = np.where(mask)[0]
    domain_ts = ts[mask]
    domain_sub = sub[mask][0]
    
    print(f"\n--- 域 ID={domain_id} (sub={domain_sub}) ---")
    print(f"样本数量: {len(indices)}")
    print(f"数组索引范围: [{indices[0]}, {indices[-1]}]")
    print(f"时间戳范围: [{domain_ts[0]:.1f}, {domain_ts[-1]:.1f}] 秒")
    print(f"前5个样本:")
    for i in range(min(5, len(indices))):
        print(f"  数组索引={indices[i]}, 时间戳={domain_ts[i]:.1f}s")
    
    # 验证是否按时间戳递增
    is_sorted = np.all(np.diff(domain_ts) >= 0)
    print(f"按时间戳递增: {'是' if is_sorted else '否'}")

# 查看某个具体被试的所有域
target_sub = 3
sub_mask = sub == target_sub
sub_domains = np.unique(d[sub_mask])
print(f"\n=== 被试 {target_sub} 的域 ===")
print(f"域列表: {sub_domains}")
```

```python
"""
分析 H5 文件中样本的时间戳顺序，找出排序问题的根源
"""
import h5py
import numpy as np
import glob
input_dir ="/vePFS-0x0d/home/zjm/de_output_addtime_new"
#input_dir = "/vePFS-0x0d/DailyEEG/de_output_addtime_new"

# 选一个有问题的被试来分析
target_sub =3
h5_path = f"{input_dir}/sub_{target_sub:02d}_de.h5"

print(f"分析 H5 文件: {h5_path}")

with h5py.File(h5_path, "r") as hf:
    trial_keys = sorted(hf.keys(), key=lambda k: int(k.replace("trial", "")))
    print(f"\nTrial 数量: {len(trial_keys)}")
    
    # 收集每个 trial 的时间戳范围和样本数量
    trial_info = []
    for trial_key in trial_keys:
        trial_idx = int(trial_key.replace("trial", ""))
        trial_grp = hf[trial_key]
        sample_keys = sorted(trial_grp.keys(), key=lambda k: int(k.replace("sample", "")))
        
        timestamps = []
        for sample_key in sample_keys:
            sample_grp = trial_grp[sample_key]
            if "raw_eeg_10s_window" in sample_grp:
                start_sec = float(sample_grp.attrs.get("start_sec", 0.0))
                timestamps.append(start_sec)
        
        if timestamps:
            day_id = int(trial_grp[sample_keys[0]].attrs["day_id"])
            trial_info.append({
                'trial_idx': trial_idx,
                'day_id': day_id,
                'n_samples': len(timestamps),
                'ts_min': min(timestamps),
                'ts_max': max(timestamps),
                'ts_list': timestamps
            })
    
    # 按域分组分析
    domains = {}
    for info in trial_info:
        if info['day_id'] not in domains:
            domains[info['day_id']] = []
        domains[info['day_id']].append(info)
    
    # 分析每个域的时间戳顺序
    for day_id in sorted(domains.keys())[:3]:  # 只看前3个域
        trials = domains[day_id]
        print(f"\n=== 域 day_id={day_id} ===")
        print(f"包含 {len(trials)} 个 trial")
        
        # 显示每个 trial 的时间戳范围
        for t in trials:
            print(f"  trial{t['trial_idx']}: {t['n_samples']}个样本, 时间戳 [{t['ts_min']:.1f}, {t['ts_max']:.1f}]")
        
        # 检查整体时间戳顺序
        all_ts = []
        for t in trials:
            all_ts.extend(t['ts_list'])
        
        # 按当前顺序排列的时间戳
        print("\n  当前顺序的时间戳（每5个样本显示一个）:")
        for i in range(0, len(all_ts), 1):
            print(f"    索引{i}: {all_ts[i]:.1f}s")
        
        # 检查是否递增
        is_sorted = np.all(np.diff(all_ts) >= 0)
        print(f"\n  按当前顺序是否递增: {'是' if is_sorted else '否'}")
        
        # 找出时间戳下降的位置
        if not is_sorted:
            diffs = np.diff(all_ts)
            drop_positions = np.where(diffs < 0)[0]
            print(f"  时间戳下降的位置（前5个）:")
            for pos in drop_positions[:5]:
                print(f"    索引{pos}→{pos+1}: {all_ts[pos]:.1f}s → {all_ts[pos+1]:.1f}s")
        
        # 如果按时间戳重新排序
        sorted_indices = np.argsort(all_ts)
        sorted_ts = np.array(all_ts)[sorted_indices]
        print(f"\n  按时间戳排序后的前10个时间戳:")
        for i in range(min(10, len(sorted_ts))):
            print(f"    {sorted_ts[i]:.1f}s")
```
